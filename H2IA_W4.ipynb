# Resolver o quebra-cabeças usando Buscas

## Estrutura

```python
# tupla de tuplas 3x3
OBJETIVO = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

class No:
    def __init__(self, estado, pai=None, acao=None, profundidade=0):
        self.estado = estado
        self.pai = pai
        self.acao = acao
        self.profundidade = profundidade

def encontrar_zero(estado):
    for i in range(3):
        for j in range(3):
            if estado[i][j] == 0:
                return i, j

def gerar_filhos(no):
    filhos = []
    linha, coluna = encontrar_zero(no.estado)

    acoes = [
        ("CIMA", -1, 0),
        ("BAIXO", 1, 0),
        ("ESQUERDA", 0, -1),
        ("DIREITA", 0, 1)
    ]

    for nome_acao, delta_linha, delta_coluna in acoes:
        nova_linha = linha + delta_linha
        nova_coluna = coluna + delta_coluna

        if 0 <= nova_linha < 3 and 0 <= nova_coluna < 3:

            # converte a tupla atual em lista de listas para poder alterar
            novo_estado = [list(row) for row in no.estado]

            novo_estado[linha][coluna], novo_estado[nova_linha][nova_coluna] = novo_estado[nova_linha][nova_coluna], novo_estado[linha][coluna]

            # converte a lista de listas de volta para tupla de tuplas
            estado_final_lista = []
            for row in novo_estado:
                estado_final_lista.append(tuple(row)) # converte cada linha em tupla
            estado_final_tupla = tuple(estado_final_lista) # converte o bloco todo em tupla

            filhos.append(No(estado_final_tupla, no, nome_acao, no.profundidade + 1))

    return filhos

def reconstituir_caminho(no):
    caminho = []
    while no:
        caminho.append((no.estado, no.acao))
        no = no.pai
    caminho.reverse()
    return caminho

```

## Busca em largura

```python
def busca_em_largura(estado_inicial):

    no_inicial = No(estado_inicial)

    if no_inicial.estado == OBJETIVO:
        return [(no_inicial.estado, None)]

    fronteira = [no_inicial]

    visitados = set()
    visitados.add(no_inicial.estado)

    while len(fronteira) > 0:

        no_atual = fronteira.pop(0)

        filhos_gerados = gerar_filhos(no_atual)

        for filho in filhos_gerados:

            if filho.estado == OBJETIVO:
                print(f"Visitados: {len(visitados)}")
                print(f"Fronteira: {len(fronteira)}")
                return reconstituir_caminho(filho)

            if filho.estado not in visitados:
                visitados.add(filho.estado)
                fronteira.append(filho)

    return None
```

## Busca em Profundidade

```python
def busca_em_profundidade(estado_inicial):

    no_inicial = No(estado_inicial)

    if no_inicial.estado == OBJETIVO:
        return [(no_inicial.estado, None)]

    fronteira = [no_inicial]

    visitados = set()
    visitados.add(no_inicial.estado)

    while len(fronteira) > 0:

        no_atual = fronteira.pop()

        filhos_gerados = gerar_filhos(no_atual)

        for filho in filhos_gerados:

            if filho.estado == OBJETIVO:
                print(f"Visitados: {len(visitados)}")
                print(f"Fronteira: {len(fronteira)}")
                return reconstituir_caminho(filho)

            if filho.estado not in visitados:
                visitados.add(filho.estado)
                fronteira.append(filho)

    return None
```

## Teste BFS

```python
import time

tabuleiro_teste = (
    (8, 2, 3),
    (1, 6, 4),
    (7, 0, 5)
)

print("Iniciando BFS")
inicio = time.time()
caminho_solucao = busca_em_largura(tabuleiro_teste)
fim = time.time()

tempo_execucao = fim - inicio
print(f"Tempo de execução do algoritmo: {tempo_execucao:.5f} segundos")

if caminho_solucao:
     print(f"Solução encontrada em {len(caminho_solucao) - 1} movimentos\n")
    # for passo, (estado, acao) in enumerate(caminho_solucao):
    #     if acao is None:
    #         print("Estado inicial:")
    #     else:
    #         print(f"Passo {passo} - Movimento: {acao}")

    #     for linha in estado:
    #         print(linha)
    #     print("-" * 20)
else:
    print("Sem solução.")
```

## Discorra sobre o desempenho dos métodos em questões de:


1.   Consumo de memória
2.   Processamento



O tempo de processamento medido em segundos varia de acordo com o Hardware. Apesar disso, através da quantidade de nós visitados podemos ter uma noção sólida da diferença de esforço computacional, uma vez que o tempo pra acessar cada nó é o mesmo. Nesse cenário, BFS processou muito menos nós para encontrar a solução do que o DFS.

Em relação a solução, BFS é um algoritm ótimo, explorando o grafo em largura ele garantiu o caminho mais curto possível, enquanto o DFS foi afundando em caminhos sem saída e retornando uma solução impraticável

O uso de memória pode ser analisado pelo tamanho da fronteira. O BFS teoricamente consome mais memória devido à explosão combinatória de guardar todos os nós do mesmo nível, porém como o DFS não tinha um limite de profundidade, pra natureza específica desse problema, ele acaba acumulando nós de forma excessiva na pilha.
